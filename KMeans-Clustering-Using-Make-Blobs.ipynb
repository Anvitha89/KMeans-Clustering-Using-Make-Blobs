```python
from sklearn.datasets import make_blobs
import numpy as np
from sklearn.metrics import silhouette_score
from sklearn.cluster import KMeans
```

```python
#Dataset Creation
blob_centers = np.array([[2.5, 3.4], [1.2, 2.4], [4.1, 3.6], [3.1, 2.1], [1.1, 2]])
blob_std = np.array([0.5, 0.4, 0.2, 0.2, 0.2])
```

```python
X, y = make_blobs(n_samples = 2000, centers = blob_centers, cluster_std = blob_std, random_state = 43)
```

```python
#KMeans Clustering
kmeans = KMeans(n_clusters = 2, n_init = 20, random_state = 7)
```

```python
#Silhouette Score
kmeans = KMeans(n_clusters = 2, n_init = 20, random_state = 6)
y_pred = kmeans.fit_predict(X)
sil_score2 = silhouette_score(X, kmeans.labels_)
print(sil_score2)
```

```python
kmeans = KMeans(n_clusters = 3, n_init = 20, random_state = 6)
y_pred = kmeans.fit_predict(X)
sil_score3 = silhouette_score(X, kmeans.labels_)
print(sil_score3)
```

```python
kmeans = KMeans(n_clusters = 4, n_init = 20,  random_state = 6)
y_pred = kmeans.fit_predict(X)
sil_score4 = silhouette_score(X, kmeans.labels_)
print(sil_score4)
```

```python
kmeans = KMeans(n_clusters = 5, n_init = 20, random_state = 6)
y_pred = kmeans.fit_predict(X)
sil_score5 = silhouette_score(X, kmeans.labels_)
print(sil_score5)
```
